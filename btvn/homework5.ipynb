### Viết hàm hoán đổi hai số nguyên. def swap(a, b) -> Tuple[int, int]: 

```python
def swap(a, b):
    temp = a
    a = b
    b = temp
    return a, b

x = 5
y = 10
x, y = swap(x, y)
print(x, y)
```

### Viết hàm kiểm tra một số có phải là số nguyên tố hay không? 


```python
def Is_Prime_Fast(n: int) -> bool:
    if n < 2: return False
    if n == 2: return True
    if n % 2 == 0: return False 
    
    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
    return True
```

#### W5A4
### VIết hàm kiểm tra một số có phải số hoàn hảo hay không, xử lý trên dữ liệu kiểu số nguyên. 


```python
def is_perfect(n):
    if n <= 0:
        return False
    
    tong = 0
    for i in range(1, n):
        if n % i == 0:
            tong = tong + i
            
    return tong == n

print(is_perfect(6))
print(is_perfect(28))
print(is_perfect(10))
```

### Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list số nguyên, nếu không có thì trả về -1 


```python
def find_first_index(nums, k):
    for i in range(len(nums)):
        if nums[i] == k:
            return i
    return -1


lst = [int(x) for x in input().split()]
k = int(input())
print(find_first_index(lst, k))
```

### Viết hàm giả lập một máy tính bỏ túi. 
### Hàm thực hiện tính toán kết quả của phép toán giữa số \( num1 \) và \( num2 \) với toán tử \( operat \) 
### Biết các toán tử đơn giản thực hiện phép toán lần lượt là: 
### Phép trừ \( - \) 
### Phép cộng \( + \) 
### Phép chia \( / \) 
### Phép nhân \( * \) 
### Kết quả làm tròn đến chữ số thập phân thứ hai. 
### Đầu vào: Hai số thực và phép toán cần thực hiện (trong số các phép toán \(+, -, *, /\)). 
### Đầu ra: Kết quả phép toán. 

```python
def calculator(num1, num2, operator):
    if operator == '+':
        result = num1 + num2
    elif operator == '-':
        result = num1 - num2
    elif operator == '*':
        result = num1 * num2
    elif operator == '/':
        if num2 == 0:
            return "Error"
        result = num1 / num2
    else:
        return None
        
    return round(result, 2)
n1 = float(input())
n2 = float(input())
op = input().strip()

print(calculator(n1, n2, op))
```

#### W5A7
### Viết hàm tính giai thừa. Nhập vào số nguyên n, trả về giai thừa của n (n!) 

```python
def tinh_giai_thua(n: int) -> int:
    if n < 0:
        return "Không tính được giai thừa của số âm"
    
    ket_qua = 1
    for i in range(1, n + 1):
        ket_qua *= i
    return ket_qua

n = int(input("Nhập n: "))
print(f"{n}! = {tinh_giai_thua(n)}")
```

#### W5A8
### Khoảng cách Hamming giữa hai số nguyên là số lượng vị trí khác nhau giữa hai dãy bits tương ứng của chúng. Ví dụ: Khoảng cách Hamming giữa 1 (1) và 4 (100) là 2. Cho hai số nguyên x và y, hãy viết hàm trả về khoảng cách Hamming giữa hai số này. Khoảng cách Hamming là cái tên được đặt theo tên của Richard Hamming, người giới thiệu lý thuyết này trong tài liệu có tính cơ sở của ông về mã phát hiện lỗi và sửa lỗi (error-detecting and error-correcting codes). Nó được sử dụng trong kỹ thuật viễn thông để tính số lượng các bit trong một từ nhị phân (binary word) bị đổi ngược, như một hình thức để ước tính số lỗi xảy ra trong quá trình truyền thông, và vì thế, đôi khi, nó còn được gọi là khoảng cách tín hiệu (signal distance). Việc phân tích trọng số Hamming của các bit còn được sử dụng trong một số ngành, bao gồm lý thuyết tin học, lý thuyết mã hóa, và mật mã học. Tuy vậy, khi so sánh các dãy ký tự có chiều dài khác nhau, hay các dãy ký tự có xu hướng không chỉ bị thay thế không thôi, mà còn bị ảnh hưởng bởi dữ liệu bị lồng thêm vào, hoặc bị xóa đi, phương pháp đo lường phức tạp hơn, như khoảng cách Levenshtein (Levenshtein distance) là một phương pháp có tác dụng và thích hợp hơn. 

```python
def hamming_distance(x, y):
    xor_result = x ^ y
    distance = 0
    while xor_result > 0:
        distance += xor_result & 1
        xor_result >>= 1
    return distance

x = int(input())
y = int(input())
print(hamming_distance(x, y))
```

#### W5A9
### Viết hàm trả về tổng các chữ số của số nguyên dương n 

```python
def sum_digits(n):
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total

n = int(input())
print(sum_digits(n))
```

#### W5A10
### Hai từ được gọi là đẳng cấu nếu các kí tự trong một từ có thể được ánh xạ để được từ còn lại. Ánh xạ một kí tự nghĩa là thay đổi toàn bộ các kí tự đó trong từ bởi một kí tự khác. Thứ tự của các kí tự được giữ không đổi. Hai kí tự khác nhau không thể ánh xạ đến cùng một kí tự, nhưng một kí tự có thể ánh xạ thành chính nó. Ví dụ, từ "abca" và "zbxz" là hai từ đẳng cấu vì ta có thể ánh xạ 'a' thành 'z', 'b' thành 'b' và 'c' thành 'x'. Viết hàm trả về \(true\) nếu hai từ \(a\) và \(b\) là đẳng cấu, ngược lại \(false\). Bạn có thể giả sử hai từ đầu vào có cùng độ dài. 

```python
def is_isomorphic(a, b):
    for i in range(len(a)):
        if a.find(a[i]) != b.find(b[i]):
            return False
    return True

word1 = input().strip()
word2 = input().strip()

print(is_isomorphic(word1, word2))
```

#### W5A12
### Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) sao cho dividend = divisor*q + r và 0 <= |r| < |divisor|. 


```python
def divide_with_remainder(dividend, divisor):
    if divisor == 0:
        return "Error"
    
    q = dividend // divisor
    r = dividend % divisor
    
    return (q, r)

dividend = int(input())
divisor = int(input())

print(divide_with_remainder(dividend, divisor))
```

#### W5A14
### Viết hàm compute(a, b, op) thực hiện phép tính a (+|-|*|/) b tùy theo ký tự op. 

```python
def compute(a, b, op):
    if op == '+':
        return a + b
    elif op == '-':
        return a - b
    elif op == '*':
        return a * b
    elif op == '/':
        if b == 0:
            return "Error"
        return a / b
    else:
        return None

a = float(input())
b = float(input())
op = input().strip()

print(compute(a, b, op))
```

### Viết hàm is_right_triangle(a, b, c) trả về True nếu ba số có thể là ba cạnh của tam giác vuông

```python
def is_right_triangle(a, b, c):
    if a + b <= c or a + c <= b or b + c <= a:
        return False
    return (a * a + b * b == c * c) or (a * a + c * c == b * b) or (b * b + c * c == a * a)   
```
