```python
class Shape:
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * self.r * self.r

try:
    loai_hinh = input().strip().upper()
    tham_so = list(map(float, input().split()))
    hinh = None

    if loai_hinh == "RECTANGLE" and len(tham_so) >= 2:
        hinh = Rectangle(tham_so[0], tham_so[1])
    elif loai_hinh == "CIRCLE" and len(tham_so) >= 1:
        hinh = Circle(tham_so[0])

    if hinh:
        print(f"{hinh.area():.2f}")
except:
    pass
```
