#### W4A19
### Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 2^k hay không 

```python
n = int(input())

bien_chay = 1

while bien_chay < n:
    bien_chay = bien_chay * 2

if bien_chay == n:
    print(f"{n} là số dạng 2^k")
else:
    print(f"{n} không phải là số dạng 2^k")
```

#### W4A20
### Nhập vào số nguyên dương a và b, in toàn bộ ước chung của a và b  

```python

a = int(input("Nhập số a: "))
b = int(input("Nhập số b: "))

if a < b:
    so_nho_nhat = a
else:
    so_nho_nhat = b

print(f"Các ước chung của {a} và {b} là:")
for i in range(1, so_nho_nhat + 1):
    if a % i == 0 and b % i == 0:
        print(i, end=" ")
```

#### W4A21
###  Nhập vào số nguyên dương n, tính tổng các chữ số của n

```python
n = int(input("Nhập vào số nguyên dương n: "))

tong = 0
while n > 0:
    chu_so_cuoi = n % 10
    tong = tong + chu_so_cuoi
    n = n // 10

print(f"Tổng các chữ số của {n} là: {tong}")
```

#### W4A22
###  Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ 


```python
n_chuoi = input("Nhập vào số nguyên dương n: ")
dem_chan = 0
dem_le = 0
for chu_so in n_chuoi:
    so = int(chu_so)
    
    if so % 2 == 0:
        dem_chan = dem_chan + 1
    else:
        dem_le = dem_le + 1

print("Số chữ số chẵn là:", dem_chan)
print("Số chữ số lẻ là:", dem_le)
```

#### W4A23
### Nhập n. Cho S(k) = 1 + 2 + 3 + … + k .Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n 

```python
n = int(input("Nhập vào số nguyên dương n: "))
tong = 0
k = 0

while tong + (k + 1) < n:
    k = k + 1      
    tong = tong + k  

print(f"Giá trị k lớn nhất tìm được là: {k}")
print(f"Khi đó tổng S({k}) = {tong} (nhỏ hơn {n})")
```

#### W4A24
### Nhập vào A, tìm n nhỏ nhất sao cho 1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A 


```python
A = float(input("Nhập vào số A: "))

tong = 0
n = 0

while tong <= A:
    n = n + 1          
    tong = tong + 1/n  

print(f"Số n nhỏ nhất tìm được là: {n}")
print(f"Khi đó tổng là: {tong:.5f} (đã lớn hơn {A})")
```

#### W4A25
### Nhập vào một dãy số nguyên, ngưng nhập khi người dùng nhập -1. Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập

```python

n = int(input("Nhập số đầu tiên (nhập -1 để dừng): "))

if n == -1:
    print("Bạn chưa nhập số nào!")
else:
    lon_nhat = n
    nho_nhat = n
    while True:
        n = int(input("Nhập số tiếp theo (nhập -1 để dừng): "))
        if n == -1:
            break 
        if n > lon_nhat:
            lon_nhat = n 
            
        if n < nho_nhat:
            nho_nhat = n 
    print("Số lớn nhất là:", lon_nhat)
    print("Số nhỏ nhất là:", nho_nhat)
```

#### W4A26
### Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ bằng tổng hai số trước đó .Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt quá A 

```python
A = int(input("Nhập vào số A: "))

so_nho = 1
so_lon = 1
while so_lon <= A:
    so_moi = so_nho + so_lon
    so_nho = so_lon
    so_lon = so_moi
print("Số Fibonacci lớn nhất không vượt quá A là:", so_nho)
```

### Nhập vào một chuỗi có dạng 3 số nguyên, mỗi số nguyên cách nhau một dấu phẩy, hãy tính tổng 3 số nguyên đó VD: Nhập: 3, 12, 15 Tổng: 30 


```python
chuoi = input("Nhập 3 số nguyên cách nhau bằng dấu phẩy: ")
danh_sach = chuoi.split(",")

tong = 0

for s in danh_sach:
    tong = tong + int(s) 

print("Tổng:", tong)
```
