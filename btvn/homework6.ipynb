#### W6A17
### Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và đường chéo phụ của mảng 

```python
n = int(input("Nhập n: "))

matrix = []
for i in range(n):
    row = []
    for j in range(n):
        value = int(input(f"Ô [{i}][{j}]: "))
        row.append(value)
    matrix.append(row)

print("Ma trận:")
for row in matrix:
    print(row)

chinh = []
phu = []

for i in range(n):
    chinh.append(matrix[i][i])
    phu.append(matrix[i][n - 1 - i])

print("Đường chéo chính:", chinh)
print("Đường chéo phụ:", phu)
```

#### W6A18
### Nhập vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột k (hàng k)

```python
n, m, k = list(map(int, input("Nhập n, m, k: ").split()))

matrix = []
for i in range(n):
    row = list(map(int, input(f"Nhập hàng {i}: ").split()))
    matrix.append(row)

tong_hang_k = sum(matrix[k])

tong_cot_k = 0
for i in range(n):
    tong_cot_k = tong_cot_k + matrix[i][k]

print("Tổng hàng k:", tong_hang_k)
print("Tổng cột k:", tong_cot_k)
```

#### W6A19
### Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất trong list

```python
def tim_vi_tri_lon_nhat(lst):
    if not lst:  # Kiểm tra nếu danh sách rỗng
        return None
   
    gia_tri_lon_nhat = max(lst)
    
   
    vi_tri = lst.index(gia_tri_lon_nhat)
    
    return vi_tri


numbers = [10, 5, 20, 15, 20]
print(f"Vị trí lớn nhất là: {tim_vi_tri_lon_nhat(numbers)}") 
```

#### W6A20
### Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về -1 

```python
def tim_vi_tri_k(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    return -1

# Chạy thử
my_list = [10, 25, 7, 30, 7, 40]
k_value = 7
print(tim_vi_tri_k(my_list, k_value))
```

#### W6A21
### Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện 

```python
def tao_dict_vi_tri(line):
    words = line.split()
    d = {}
    
    for i, word in enumerate(words):
        if word not in d:
            d[word] = []
        d[word].append(i)
    
    for word in d:
        d[word] = tuple(d[word])
        
    return d

input_line = input("Nhập chuỗi: ")
ket_qua = tao_dict_vi_tri(input_line)
print(ket_qua)
```

#### W6A22
### Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất trong list 

```python
def tim_vi_tri_max(lst):
    if not lst:
        return None
    
    vi_tri_max = 0
    for i in range(1, len(lst)):
        if lst[i] > lst[vi_tri_max]:
            vi_tri_max = i
            
    return vi_tri_max

numbers = [10, 5, 20, 15, 20]
print(tim_vi_tri_max(numbers))
```

#### W6A24
### Viết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9 

```python
def day_so_0_ve_dau(lst):
    so_0 = [x for x in lst if x == 0]
    so_khac = [x for x in lst if x != 0]
    return so_0 + so_khac


numbers = [0, 2, 3, 2, 4, 0, 0, 0, 9]
print(day_so_0_ve_dau(numbers))
```

#### W6A25
### Cho dãy chỉ gồm N giá trị từ 0 - 9 . Kiểm tra xem giá trị nào không có trong mảng

```python
def tim_gia_tri_thieu(lst):
    tap_hop_day_du = set(range(10))
    tap_hop_dau_vao = set(lst)
    ket_qua = tap_hop_day_du - tap_hop_dau_vao
    return sorted(list(ket_qua))

numbers = [0, 2, 3, 2, 4, 0, 0, 0, 9]
print(tim_gia_tri_thieu(numbers))
```

#### W6A26
### Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 phần tử đầu và cuối 

```python
def tong_cuc_dai_dia_phuong(lst):
    tong = 0
    for i in range(1, len(lst) - 1):
        if lst[i] > lst[i-1] and lst[i] > lst[i+1]:
            tong += lst[i]
    return tong

numbers = [1, 5, 2, 6, 3, 8, 4]
print(tong_cuc_dai_dia_phuong(numbers))
```

#### W6A27
### Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước). nếu không có in ra -1

```python
def tim_vi_tri_nho_hon_x(lst, x):
    vi_tri = [i for i, val in enumerate(lst) if val < x]
    
    if not vi_tri:
        print(-1)
    else:
        for v in vi_tri:
            print(v, end=" ")
        print()

numbers = [10, 5, 8, 20, 3, 15]
x_value = 10
tim_vi_tri_nho_hon_x(numbers, x_value)
```

#### W6A28
### Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm thực sự, hay vô danh (không tăng, không giảm)

```python
def kiem_tra_day_so(lst):
    if len(lst) < 2:
        return "Vô danh"

    tang = True
    giam = True

    for i in range(1, len(lst)):
        if lst[i] <= lst[i-1]:
            tang = False
        if lst[i] >= lst[i-1]:
            giam = False

    if tang:
        return "Tăng thực sự"
    elif giam:
        return "Giảm thực sự"
    else:
        return "Vô danh"

print(kiem_tra_day_so([1, 3, 5, 8]))    
print(kiem_tra_day_so([10, 7, 4, 2]))   
print(kiem_tra_day_so([1, 5, 2, 4]))   
print(kiem_tra_day_so([1, 1, 2]))       
```
