```python
def check_beautiful_array(arr):
    counts = {}
    for x in arr:
        if x in counts:
            counts[x] += 1
        else:
            counts[x] = 1
            
    y = 0
    for x in counts:
        f = counts[x]
        if f > x:
            y += (f - x)
        elif f < x:
            y += f
            
    return y

n = int(input())
arr = list(map(int, input().split()))

result = check_beautiful_array(arr)
print(result)
```

```python
def find_majority_element(arr):
    n = len(arr)
    if n == 0:
        return

    candidate = None
    count = 0

    for x in arr:
        if count == 0:
            candidate = x
            count = 1
        elif x == candidate:
            count += 1
        else:
            count -= 1

    actual_count = 0
    for x in arr:
        if x == candidate:
            actual_count += 1

    if actual_count > n / 2:
        print(candidate)

n_input = int(input())
if n_input > 0:
    arr_input = list(map(int, input().split()))
    find_majority_element(arr_input)
```

```python
def normalize(arr):
    if len(arr) > 1 and arr[0] > arr[-1]:
        return arr[::-1]
    return arr

def merge_sorted_arrays(a, b):
    a = normalize(a)
    b = normalize(b)
    
    result = []
    i = 0
    j = 0
    
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
    
    result.extend(a[i:])
    result.extend(b[j:])
    
    return result

a = list(map(int, input().split()))
b = list(map(int, input().split()))

print(*(merge_sorted_arrays(a, b)))
```

```python
n = int(input())
counts = {}

for _ in range(n):
    name = input().strip()
    if name in counts:
        counts[name] += 1
    else:
        counts[name] = 1

best_name = ""
max_freq = 0

for name in counts:
    f = counts[name]
    if f > max_freq:
        max_freq = f
        best_name = name
    elif f == max_freq:
        if name < best_name:
            best_name = name

print(best_name)
```

```python
n = int(input())

def la_so_nguyen_to(m):
    if m < 2:
        return False
    for i in range(2, int(m**0.5) + 1):
        if m % i == 0:
            return False
    return True

cac_so_nt = []
for i in range(1, n + 1):
    if la_so_nguyen_to(i):
        cac_so_nt.append(i)

max_dai = 0
bat_dau_day = -1

for i in range(len(cac_so_nt) - 1):
    p1 = cac_so_nt[i]
    p2 = cac_so_nt[i+1]
    
    do_dai_hien_tai = p2 - p1 - 1
    
    if do_dai_hien_tai > max_dai:
        max_dai = do_dai_hien_tai
        bat_dau_day = p1 + 1

if bat_dau_day != -1 and max_dai > 0:
    for i in range(max_dai):
        if i == max_dai - 1:
            print(bat_dau_day + i)
        else:
            print(bat_dau_day + i, end=" ")
```

```python
def in_day_con_tang_dai_nhat():
    arr = list(map(int, input().split()))
    n = len(arr)
    if n == 0: return

    L = [1] * n

    for i in range(1, n):
        for j in range(i):
            if arr[i] > arr[j]:
                L[i] = max(L[i], L[j] + 1)

    max_len = max(L)
    print("Độ dài:", max_len)

    result = []
    current_len = max_len
    
    for i in range(n - 1, -1, -1):
        if L[i] == current_len:
            if not result or arr[i] < result[-1]:
                result.append(arr[i])
                current_len -= 1

    print("Dãy con:", *result[::-1])

in_day_con_tang_dai_nhat()
```
