### B4
####  Viết chương trình in ra hình tam giác cân có độ cao h được nhập từ bàn phím 

```python
h = int(input("Nhập độ cao h: "))

for i in range(1, h + 1):

    spaces = " " * (h - i)
    if i == 1: 
        print(spaces + "*")
    elif i == h:
        print("*" * (2 * h - 1))
    else:
        mid_spaces = " " * (2 * i - 3)
        print(spaces + "*" + mid_spaces + "*")
```

### B9
####  Nhập vào số nguyên dương a, kiểm tra xem a có phải là số chính phương hay không? Số chính phương là số có căn bậc hai là một số nguyên VD: 1, 4, 9, 16, 25…

```python
a = int(input("Nhập vào số nguyên dương a: "))

la_so_chinh_phuong = False  

for i in range(1, a + 1):
    if i * i == a:
        la_so_chinh_phuong = True
        break 

if la_so_chinh_phuong:
    print(a, "là số chính phương")
else:
    print(a, "không phải là số chính phương")
```

### B10
#### : Nhập vào số nguyên dương a, kiểm tra xem a có phải là số hoàn hảo hay không. 
#### Số hoàn hảo là số có tổng các ước (ước nhỏ hơn nó) bằng chính nó, VD: số 6, 28… 


```python
a = int(input("Nhập vào số nguyên dương a: "))

tong_cac_uoc = 0
i = 1

while i < a:
    if a % i == 0:
        tong_cac_uoc = tong_cac_uoc + i
    i = i + 1  

if tong_cac_uoc == a:
    print(a, "là số hoàn hảo")
else:
    print(a, "không phải là số hoàn hảo")
```

### B12
#### Nhập số nguyên dương n, Cho S(k) = 1 + 2 + 3 + … + k. Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n 


```python
n = int(input("Nhập số nguyên dương n: "))

tong = 0
k = 0

while tong + (k + 1) < n:
    k = k + 1      
    tong = tong + k 

print("Số k lớn nhất để S(k) < n là:", k)
print("Tổng S(" + str(k) + ") là:", tong)
```

```python
n = int(input("Nhập số nguyên dương n: "))

tong = 0

for i in range(1, n + 1):
    if tong + i < n:
        tong = tong + i
        k = i  
    else:
        break

print("Số k lớn nhất là:", k)
print("Tổng S(" + str(k) + ") =", tong)
```

### B13
#### Nhập vào số nguyên dương A, tìm n nhỏ nhất sao cho 
#### 1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A 


```python
A = float(input("Nhập vào số nguyên dương A: "))

tong = 0
n = 0

while tong <= A:
    n = n + 1
    tong = tong + 1/n

print("Số n nhỏ nhất là:", n)
print("Tổng tương ứng là:", tong)
```

### B16
#### Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ

```python

n = int(input("Nhập vào số nguyên dương n: "))

dem_chan = 0
dem_le = 0

if n == 0:
    dem_chan = 1
else:
    temp = n 
    while temp > 0:
        chu_so = temp % 10  
        
        if chu_so % 2 == 0:
            dem_chan = dem_chan + 1
        else:
            dem_le = dem_le + 1
            
        temp = temp // 10  
        
# Bước 4: In kết quả
print("Số chữ số chẵn là:", dem_chan)
print("Số chữ số lẻ là:", dem_le)
```

### B18
#### Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 3^k hay không

```python
n = int(input("Nhập vào số nguyên dương n: "))
x = n 
if n <= 0:
    print(n, "không phải là dạng 3^k")
else:
    while n % 3 == 0 and n > 1:
        n = n // 3
    if n == 1:
        print(x, "là số dạng 3^k")
    else:
        print(x, "không phải là dạng 3^k")
```

### B19
#### Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt quá A 


```python
A = int(input("Nhập vào số nguyên dương A: "))
f1 = 1
f2 = 1
if A < 1:
    print("Không có số nào thỏa mãn.")
elif A == 1:
    print("Số Fibonacci lớn nhất không vượt quá A là: 1")
else:
    for i in range(A):
        f_tiep_theo = f1 + f2
        if f_tiep_theo > A:
            break
            
        f1 = f2
        f2 = f_tiep_theo

    print("Số Fibonacci lớn nhất không vượt quá", A, "là:", f2)
```
