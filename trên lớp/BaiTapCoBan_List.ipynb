### Bài 2b: Tìm số nhỏ nhất của 1 list và vị trí của nó, nếu có nhiều số nhỏ nhất bằng nhau thì lấy chỉ số đầu tiên (nhỏ nhất)
10 0 0 12 34 5 6 7  8 89 
0 1

```python
def min_and_pos(lst):
    if lst:
        min_val = lst[0]
        min_pos = 0
        for i in range(1, len(lst)):
            if lst[i] < min_val:
                min_val = lst[i]
                min_pos = i
        return min_val, min_pos
    return None
assert min_and_pos([3, 1, 4, 1, 5]) == (1, 1)
assert min_and_pos([]) == None
print("pass all test")
```

### Bài 3: Tìm giá trị trung bình các số nguyên dương trong list, nếu không có số nguyên dương nào trả về "invalid"

```python
def average_lst(lst):
    if lst:
        lst_positive = []
        for x in lst:
            if x > 0:
                lst_positive.append(x)
        if not lst_positive:
            return None
        t = 0
        for x in lst_positive:
            t += x
        return round(t / len(lst_positive),2)
    return "Invalid"
assert average_lst([1, -2, 3, 5]) == 3
assert average_lst([-1, -2, -3]) == None
print("pass all test!")
```

### Bài 5: Đưa tất cả các số bằng 0 trong list về đầu của list. Các số khác không thay đổi thứ tự trước sau
3 4 6 7 1 9 0 0 23 0 0 34 67 89
0 0 0 0 3 4 6 7 1 9 23 34 67 89

```python
def move_zero_to_first(lst):
    if lst is None:
        return None
    
    for i in range(len(lst) - 1, -1, -1):
        if lst[i] == 0:
            lst.insert(0, lst.pop(i))
            
    return lst

assert move_zero_to_first([1,0,2,0,3]) == [0,0,1,2,3]
assert move_zero_to_first([1,2,3]) == [1,2,3]
assert move_zero_to_first([]) == []
print("pass all test!")
```

### Bài 6: Xóa tất cả các số giống nhau trong list, Các số khác giữ nguyên vị trí trước sau

```python
def unique_array(lst):
    if lst:
        ans = []
        for x in lst:
            if x not in ans:
                ans.append(x)
        return ans
    return None
assert unique_array([1,2,2,3]) == [1,2,3]
assert unique_array([]) == None
print("pass all test!")
```

### Bài 7: Kiểm tra một list số nguyên có phải là dãy tăng, dãy giảm hay vô danh
1 3 5 6 7 8 9: Tang
9 8 7 6 5 4 0: Giam
1 2 3 7 1 0 2: Vo danh
1 2 2 4 5 7: Vo danh

```python
numbers = list(map(int, input().split()))

def kiem_tra_day_so(nums):
    n = len(nums)
    if n < 2:
        return "Dãy cần ít nhất 2 phần tử"

    tang = True
    giam = True

    for i in range(n - 1):
        if nums[i] >= nums[i+1]:
            tang = False
        if nums[i] <= nums[i+1]:
            giam = False

    if tang:
        return "Tang"
    elif giam:
        return "Giam"
    else:
        return "Vo danh"

print(kiem_tra_day_so(numbers))
```

### Bài 8: Kiểm tra một list số nguyên có phải là ziczac khong
1 3 2 5 2 8 7 9 2 100: ziczac
34 2 56 7 90 20 30 7 120: ziczac

```python
def zic_zac(lst):
    if lst:
        n = len(lst)
        for i in range(1, n-1):
            if ((lst[i]-lst[i-1]) * (lst[i]-lst[i+1]) <= 0):
                return False
        return True
    return None
assert zic_zac([1,3,2,4]) == True
assert zic_zac([1,2,3]) == False
print("pass all test!")
```

### Bài 9: Tìm và in ra tất cả các vị trí của số nguyên k trong một danh sách số nguyên cho trước. Nếu không có k trong ds thì in ra "-1".
VD: k = 3, [1,3,4,4,3,3,4,3,6,1] => k xuất hiện tại các vị trí (trả về 1 list) [1, 4, 5, 7]
k = 10, [1,3,4,4,3,3,4,3,6,1] => k không xuất hiện trong ds => trả về "-1"

```python
def tim_vi_tri_k(lst, k):
    vi_tri = []
    
    for i in range(len(lst)):
        if lst[i] == k:
            vi_tri.append(i)
            
    if not vi_tri:
        return "-1"
    
    return vi_tri
assert tim_vi_tri_k([1, 3, 4, 4, 3, 3, 4, 3, 6, 1], 3) == [1, 4, 5, 7]
assert tim_vi_tri_k([1, 3, 4, 4, 3, 3, 4, 3, 6, 1], 10) == "-1"
print("pass all test!")
```

### Bài 10: 
Nhập vào tên một người, in ra kí tự đầu tiên và kí tự cuối cùng của tên đó.
VD: "Nguyen Ai Quoc" => ["N", "c"]

```python
def get_first_last_char(name):
    if not name:
        return []
    first_char = name[0]
    last_char = name[-1]
    
    return [first_char, last_char]
assert get_first_last_char("Hello") == ['H', 'o']
assert get_first_last_char("") == []
print("pass all test!")
```
