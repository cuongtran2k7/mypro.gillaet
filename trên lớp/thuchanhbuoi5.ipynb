### W4A6
#### Nhập hai số nguyên dương a, b (a ≤ b). Hãy tính tổng các số nguyên tố trong đoạn [a, b]. Mỗi dòng là một bộ test (a b).

```python

n = int(input("Số bộ test: "))

for _ in range(n):
  
    a, b = map(int, input().split())
    
    tong_snt = 0

    for num in range(a, b + 1):
        if num > 1:
            la_snt = True
            for i in range(2, num): 
                if num % i == 0:
                    la_snt = False
                    break
            
            if la_snt:
                tong_snt += num
                
    print(tong_snt)
```

### W4A7
#### Nhập vào số nguyên dương n (n ≥ 2). Hãy in ra ước số nguyên tố lớn nhất của n.

```python
n = int(input("Nhập n: "))

uoc_max = 0
for i in range(n, 1, -1):
    if n % i == 0:
        la_snt = True
        for j in range(2, i):
            if i % j == 0:
                la_snt = False
                break
        if la_snt:
            uoc_max = i
            break 

print(uoc_max)
```

### W4A8
#### Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số đối xứng (palindrome). In ra số bước và giá trị palindrome thu được. 

```python
n = int(input("Nhập một số: "))
so_buoc = 0

while True:
    chu_goc = str(n)
    chu_dao = ""
    
    for chu_cai in chu_goc:
        chu_dao = chu_cai + chu_dao
    
    if chu_goc == chu_dao:
        break
    
    n = n + int(chu_dao)
    so_buoc = so_buoc + 1

print("Số bước thực hiện là:", so_buoc)
print("Số đối xứng cuối cùng là:", n)
```

### W4A10
#### Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số nhỏ nhất

```python
n = int(input())

max_l = 0
x_min = 0

for i in range(1, n + 1):
    curr = i
    length = 1
    while curr > 1:
        if curr % 2 == 0:
            curr //= 2
        else:
            curr = 3 * curr + 1
        length += 1
    
    if length > max_l:
        max_l = length
        x_min = i

print(x_min, max_l)
```

### W4A11
#### Viết chương trình đếm số lượng ước số chẵn của số nguyên dương n với n < 10^6.

```python
n = int(input("Nhập số nguyên dương n: "))

dem = 0
for i in range(1, n + 1):
    if n % i == 0:
        if i % 2 == 0:
            dem = dem + 1

print("Số lượng ước số chẵn là:", dem)
```

### Số này bằng tổng tất cả các ước của số kia (trừ chính nó) và ngược lại. 
### Ví dụ 220 và 284 là một cặp số thân thiết 220 ngoài chính nó ra, nó còn có 11 ước số khác là: 1, 2, 4, 5, 10, 11, 20, 44, 55, 110.Tổng của 11 ước này cho chúng ta 284 
### 284 ngoài bản thân nó, còn có 5 ước số khác là: 1, 2, 4, 71, 142. Tổng của chúng vừa đúng bằng 220 
### Nhập hai số từ bàn phím và kiểm tra xem chúng có phải là cặp số thân thiết không. Nếu có, in ra màn hình "true", ngược lại, in ra "false" 


```python

def tong_uoc(n):
    tong = 0
    # Chạy vòng lặp từ 1 đến n-1
    for i in range(1, n):
        if n % i == 0:
            tong = tong + i
    return tong

a = int(input("Nhập số thứ nhất: "))
b = int(input("Nhập số thứ hai: "))

tong_a = tong_uoc(a) 
tong_b = tong_uoc(b) 

if tong_a == b and tong_b == a:
    print("true")
else:
    print("false")
```

### Từ bài toán dân gian trên, mở rộng thành chương trình nhận đầu vào là tổng số con và tổng số chân của gà và chó. Nếu tìm được số phù hợp, in ra số lượng gà và số lượng chó. Ngược lại, in ra "invalid"

```python
tong_con = int(input("Nhập tổng số con: "))
tong_chan = int(input("Nhập tổng số chân: "))

da_tim_thay = False

for ga in range(tong_con + 1):
    cho = tong_con - ga

    if (ga * 2 + cho * 4) == tong_chan:
        print(f"Số gà: {ga}")
        print(f"Số chó: {cho}")
        da_tim_thay = True
        break  

if not da_tim_thay:
    print("invalid")
```
